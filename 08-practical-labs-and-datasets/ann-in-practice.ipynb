# Artificial Neural Network (ANN): A Practical Guide

This notebook provides a practical walkthrough of building a simple Artificial Neural Network, applying it to the `ann-data.csv` dataset. It serves as the hands-on counterpart to the concepts explained in the `artificial-neural-network.md` document.

## 0. Setup and Data Loading

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Load the dataset
df = pd.read_csv('ann-data.csv')
```

## 1. Data Exploration

```python
print("--- First 5 rows of the dataset ---")
print(df.head())

print("\n--- DataFrame Info ---")
df.info()

print("\n--- Target variable distribution ---")
print(df['target'].value_counts())
```

## 2. Data Preprocessing: Feature Scaling

Neural networks are sensitive to feature scaling, so it's a critical preprocessing step. We'll use `StandardScaler` to standardize the features to have a mean of 0 and a standard deviation of 1.

```python
X = df.drop('target', axis=1)
y = df['target']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("--- First 5 rows of scaled features ---")
print(X_scaled[:5])
```

## 3. Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)
```

## 4. Building and Training the ANN

```python
# Initialize the Multi-layer Perceptron Classifier
# We'll create a simple network with two hidden layers.
mlp = MLPClassifier(hidden_layer_sizes=(10, 5), max_iter=1000, random_state=42)

# Train the model
mlp.fit(X_train, y_train)
```

## 5. Making Predictions and Evaluating the Model

With a trained model, we can now make predictions on the test set, which the model has never seen before. We then evaluate its performance by comparing the predictions to the true labels.

The **Classification Report** provides key metrics like precision, recall, and F1-score, which are useful for understanding the model's performance on a per-class basis. The **Confusion Matrix** gives a visual representation of where the model made correct and incorrect predictions.

```python
# Make predictions on the test set
y_pred = mlp.predict(X_test)

# Generate the classification report
print("--- Classification Report ---")
print(classification_report(y_test, y_pred))

# Generate the confusion matrix
cm = confusion_matrix(y_test, y_pred)

# Visualize the confusion matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=np.unique(y), yticklabels=np.unique(y))
plt.title('Confusion Matrix')
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.show()
```

## 6. Conclusion

This notebook demonstrated the process of building a simple Artificial Neural Network for a classification task. We highlighted the importance of feature scaling for ANNs and then trained and evaluated the model. The results show that even a basic neural network can achieve high performance on this dataset.
