# Decision Trees: A Practical Guide

This notebook provides a practical walkthrough of the key concepts of Decision Trees, applying them to the `decision-tree-data.csv` dataset. It serves as the hands-on counterpart to the concepts explained in the `decision-trees.md` document.

## 0. Setup and Data Loading

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('decision-tree-data.csv')
```

## 1. Data Exploration

```python
print("--- First 5 rows of the dataset ---")
print(df.head())

print("--- DataFrame Info ---")
df.info()
```

## 2. Data Preprocessing

Decision trees can handle categorical data, but the scikit-learn implementation requires all features to be numerical. We'll use `LabelEncoder` to convert our categorical text data into numbers.

```python
le = LabelEncoder()
df_encoded = df.apply(le.fit_transform)

print("--- First 5 rows of the encoded dataset ---")
print(df_encoded.head())
```

## 3. Train-Test Split

We'll split the data into features (X) and the target variable (y), and then divide it into training and testing sets.

```python
X = df_encoded.drop('play', axis=1)
y = df_encoded['play']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

## 4. Building and Training the Decision Tree

```python
# Initialize the Decision Tree Classifier
# We use 'entropy' as the criterion to align with the Information Gain calculations in the markdown.
clf = DecisionTreeClassifier(criterion='entropy', random_state=42)

# Train the model
clf.fit(X_train, y_train)
```

## 5. Visualizing the Decision Tree

```python
plt.figure(figsize=(20,10))
plot_tree(clf, 
           feature_names=X.columns, 
           class_names=[str(c) for c in le.classes_],
           filled=True, 
           rounded=True)
plt.title("Decision Tree Visualization")
plt.show()
```

## 6. Making Predictions and Evaluating the Model

After training, we use the model to make predictions on our unseen test data (`X_test`). We then compare these predictions (`y_pred`) to the actual true labels (`y_test`) to evaluate how well our model has performed.

```python
# Make predictions on the test set
y_pred = clf.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy:.2f}")

# Generate the confusion matrix
cm = confusion_matrix(y_test, y_pred)

# Visualize the confusion matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=le.classes_, yticklabels=le.classes_)
plt.title('Confusion Matrix')
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.show()
```

## 7. Conclusion

This notebook demonstrated the end-to-end process of building a decision tree classifier. We loaded the data, preprocessed it by encoding categorical variables, trained the model, visualized the resulting tree, and evaluated its performance. The visualization is particularly powerful, as it allows us to see the exact decision rules the model learned from the data.
