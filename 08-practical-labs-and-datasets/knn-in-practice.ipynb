# K-Nearest Neighbors (KNN): A Practical Guide

This notebook provides a practical walkthrough of the KNN algorithm, applying it to the knn-data.csv dataset. It serves as the hands-on counterpart to the concepts explained in the k-nearest-neighbors.md document.

## 0. Setup and Data Loading

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('knn-data.csv')
```

## 1. Data Exploration

```python
print("--- First 5 rows of the dataset ---")
print(df.head())

print("\n--- DataFrame Info ---")
df.info()
```

## 2. Data Preprocessing: Feature Scaling

Since the KNN algorithm relies on the distance between data points, it's crucial to scale the features so that all of them contribute equally to the distance calculation. We'll use StandardScaler to standardize our features to have a mean of 0 and a standard deviation of 1.

```python
scaler = StandardScaler()
X = df.drop('target', axis=1)
y = df['target']

X_scaled = scaler.fit_transform(X)

print("--- First 5 rows of scaled features ---")
print(X_scaled[:5])
```

## 3. Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)
```

## 4. Finding the Optimal K Value

A critical step in using KNN is choosing the right number of neighbors (K). We can do this by training models with different K values and plotting their error rates. The 'elbow' of the resulting curve is often the best choice for K.

```python
error_rate = []

# Will take some time
for i in range(1, 40):
    knn = KNeighborsClassifier(n_neighbors=i)
    knn.fit(X_train, y_train)
    pred_i = knn.predict(X_test)
    error_rate.append(np.mean(pred_i != y_test))

plt.figure(figsize=(10, 6))
plt.plot(range(1, 40), error_rate, color='blue', linestyle='dashed', marker='o',
         markerfacecolor='red', markersize=10)
plt.title('Error Rate vs. K Value')
plt.xlabel('K')
plt.ylabel('Error Rate')
plt.grid(True)
plt.show()
```

From the plot, we can see that the error rate is minimized for a K value around 5-7. Let's choose K=5 for our final model.

## 5. Building and Training the KNN Model (with Optimal K)

```python
# Initialize the KNN Classifier with the optimal K
knn = KNeighborsClassifier(n_neighbors=5)

# Train the model
knn.fit(X_train, y_train)
```

## 6. Making Predictions and Evaluating the Model

Now that we have a trained model with our chosen K, we can make predictions on the unseen test data. We'll then evaluate its performance using two common metrics:
- **Classification Report:** This gives us a detailed breakdown of precision, recall, and F1-score for each class, providing a deeper insight into the model's performance beyond simple accuracy.
- **Confusion Matrix:** This table visualizes the model's performance by showing the actual vs. predicted classes. It's a great way to see where the model is getting confused.

```python
# Make predictions on the test set
y_pred = knn.predict(X_test)

# Generate the classification report
print("--- Classification Report ---")
print(classification_report(y_test, y_pred))

# Generate the confusion matrix
cm = confusion_matrix(y_test, y_pred)

# Visualize the confusion matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=np.unique(y), yticklabels=np.unique(y))
plt.title('Confusion Matrix')
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.show()
```

## 7. Conclusion

This notebook demonstrated how to build a KNN classifier. We emphasized the importance of feature scaling and showed a practical method (the elbow method) for selecting an optimal value for K. The final model's performance was then evaluated using a classification report and a confusion matrix.
